```python
from gwpy.timeseries import TimeSeries
import h5py
import glob
import numpy as np
```

```python
ts = TimeSeries.fetch_open_data(
    ifo='H1',
    start=1238171648,
    end=1238175744,
    sample_rate=16384,
    format='gwf',
    host='https://gwosc.org',
)
```

```python
print(ts)
ts.write(
    "/home/jerry/data_cache/H1/H-H1_GWOSC_16KHZ_R1_STRAIN-1238171648-4096.gwf",
    format='gwf',
)
```

```python
trigger_files = glob.glob(f"/home/jerry/GW/gw-anomaly-detection/omicron/triggers/H1:GWOSC-16KHZ_R1_STRAIN/*.h5")
trigger_files = sorted(trigger_files)
trigger_infos = []
for trigger_file in trigger_files:
    with h5py.File(trigger_file, 'r') as f:
        names = f['triggers'].dtype.names
        trigger_infos.extend(f['triggers'][:])

output_file = "/home/jerry/GW/gw-anomaly-detection/omicron/triggers/merge.hdf5"
with h5py.File(output_file, 'w') as w:
    data = np.stack(trigger_infos)
    w.create_dataset(
        "glitch_info",
        shape=data.shape,
        dtype=data.dtype,
        data=data,
    )
```

```python
with h5py.File(output_file, 'r') as f:
    print(f['glitch_info'].shape)
```
